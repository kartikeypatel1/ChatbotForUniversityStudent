# EDA - University Student Support Chatbot (dataset.csv)
Run `python src/clean_and_eda.py` first (it creates `data/dataset_clean.csv`), or run the cells below. Write a short observation under each chart for your report.

```
import sys; sys.path.insert(0, '../src')
import pandas as pd, seaborn as sns, matplotlib.pyplot as plt
from clean_and_eda import load_raw, clean
sns.set_theme(style='whitegrid')
log = []
df = clean(load_raw('../data/dataset.csv'), log)
print('\n'.join(log))
df.head()
```

## 1. Data overview

```
print(df.shape)
df.dtypes
```

```
df[['confidence_score','rating','query_length','word_count']].describe()
```

## 2. Queries per intent  
*Observation: which intents dominate? Is the data balanced?*

```
plt.figure(figsize=(8,5)); sns.countplot(y='intent', data=df, order=df['intent'].value_counts().index, color='#4C72B0'); plt.title('Queries per intent'); plt.show()
```

## 3. Confidence score distribution  
*Observation: do escalated chats have lower confidence?*

```
sns.histplot(data=df.dropna(subset=['escalated']), x='confidence_score', hue='escalated', bins=20); plt.title('Confidence by escalation'); plt.show()
```

## 4. Escalation rate by intent  
*Observation: which topics need humans most often?*

```
rate = df.dropna(subset=['escalated']).assign(esc=lambda d:(d['escalated']=='Y').astype(int)).groupby('intent')['esc'].mean().sort_values()
rate.plot.barh(figsize=(8,5)); plt.xlabel('Escalation rate'); plt.show()
```

## 5. Ratings  
*Observation: are escalated chats rated lower?*

```
fig, ax = plt.subplots(1, 2, figsize=(11,4))
sns.countplot(x='rating', data=df.dropna(subset=['rating']), ax=ax[0], color='#55A868')
sns.boxplot(x='escalated', y='rating', data=df.dropna(subset=['escalated','rating']).astype({'rating': float}), ax=ax[1])
plt.show()
```

## 6. Query length  
*Observation: are longer / angrier messages escalated more?*

```
sns.boxplot(x='escalated', y='word_count', data=df.dropna(subset=['escalated'])); plt.show()
```
